The len() function counts every character in the string, including:

✅ Letters
✅ Numbers
✅ Spaces
✅ Special characters (., ,, !, @, etc.)

```python
sentences = input("Enter a sentence:")
words = sentences.split()
word_count = len(words)

character_count = len(sentences.replace(" ", ""))  # Count characters excluding spaces


vowels = "aeiouAEIOU"
vowel_count = 0
consonents_count = 0

for ch in sentences:
    if ch.isalpha():
        if ch in vowels:
            vowel_count += 1
        else:
            consonents_count += 1

print("Entered Sentence:", sentences)
print("Number of words:", word_count)
print("Number of characters:", character_count)
print("Number of vowels:", vowel_count)
print("Number of consonents:", consonents_count)
```

Problem 5. Write a program to determine whether a given string is a palindrome. Ignore spaces and case sensitivity.

```python
# Input from the user
text = input("Enter a string: ")

# Remove spaces and convert to lowercase
text = text.replace(" ", "").lower()

print("Entered String:", text)
# Check palindrome
if text == text[::-1]:
    print("The string is a palindrome.")
else:
    print("The string is not a palindrome.")
```

Problem 6. Accept a tuple containing student marks. Display Highest mark • Lowest mark • Number of students scoring above the class average

```python
marks = tuple(map(float, input("Enter marks for 5 subjects: ").split()))


Highest_mark = max(marks)
Lowest_mark = min(marks)

average = sum(marks) / len(marks)

count = 0
for mark in marks:
    if mark > average:
        count += 1
        
print("Display Marks:", marks)
print("Highest mark:", Highest_mark)
print("Lowest mark:", Lowest_mark)
print("Average mark:", average)
print("Number of subjects with marks above average:", count)
```

Problem 7. Create a dictionary where the keys are student names and the values are their marks. Allow the user to search for a student’s marks by entering the student’s name.

```python
students = {}

n = int(input("Enter the number of students:"))
for i in range(n):
    name = input("Enter the Student Name:")
    marks = float(input("Enter the Student Marks:"))
    students[name] = marks
    
print("Student Marks Dictionary:", students)
print("Student Names:", list(students.keys()))
search = input("Enter the name of the student to search:")

if search in students:
    print("marks of", search, "is", students[search])
else:
    print("Student not found.")
    
```

Problem 8. Write a program that counts the frequency of each character in a given string using a dictionary. Example: banana should produce b : 1 a : 3 n : 2

```python
text = input("Enter a string: ")

frequency = {}

for ch in text:
    if ch in frequency:
        frequency[ch] += 1
    else:
        frequency[ch] = 1

print("Character Frequency:")

for ch in frequency:
    print(ch, ":", frequency[ch])
```

Problem 9. Accept two sets from the user and display • Union • Intersection • Difference • Symmetric Difference

```python
set1 = set(map(int, input("Enter elements of Set 1: ").split()))
set2 = set(map(int, input("Enter elements of Set 2: ").split()))

union = set1 | set2
intersection = set1 & set2
difference = set1 - set2
symmetric_difference = set1 ^ set2

print("Set 1:", set1)
print("Set 2:", set2)
print("Union:", union)
print("Intersection:", intersection)
print("Difference:", difference)
print("Symmetric Difference:", symmetric_difference)
```
