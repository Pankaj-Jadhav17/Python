 List Comprehensions

```python
arr = [50, 60, 100]
list1 = [p * 2 for p in arr if p > 60]
print(list1)
```

```python
domain = ["www.google.com", "openai.com", "localhost", "www.something.com"]
cleaned = [d.replace("www.", "") for d in domain]
print(cleaned)
```

```python
only_www = [d.replace("www.", "") 
            for d in domain 
              if "www." in d]
print(only_www)
```

```python
arr = ["Alice", 23, "Bob", "100 BCE"]
for a in arr:
    if isinstance(a, str) and not any(ch.isdigit() for ch in a):
        print(a)
```

```python
arr = ["Alice", 23, "Bob", "100 BCE"]
strings_only = [a for a in arr if isinstance(a, str) and not any(ch.isdigit() for ch in a)]
print(strings_only)
```

Sets

```python
mset = {1, 2, 3, 4, 5, 6, 7}
mset.add(10)
mset.add(22)
print(mset)
```

```python
print(mset[2])   # attempt to index
```

```python
mset.remove(5)
print(mset)
```

```python
mset = {1, 2, 3, 4, 6, 7, 10, 22}
nset = {3, 4, 5, 100}
print(mset | nset)          # union
print(mset & nset)          # intersection
print({1, 2} <= mset)       # subset check
```
