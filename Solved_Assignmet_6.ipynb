Store the values in a NumPy array.
Determine:

```python
temperatures = np.array([32.5, 31.8, 30.2, 33.1, 34.5, 29.8, 31.0])
print(temperatures)

average = np.mean(temperatures)
print(average)
maximum = np.max(temperatures)
print(maximum)
minimum = np.min(temperatures)
print(minimum)
day = np.argmax(temperatures)
print(day)
day = np.argmax(temperatures) + 1
print(day)
above_average = temperatures[temperatures > average]
print(above_average)
```

```python
def temperatures_analysis(temperatures):
    average = np.mean(temperatures)
    maximum = np.max(temperatures)
    minimum = np.min(temperatures)
    highest_day = np.argmax(temperatures) + 1
    above_average = temperatures[temperatures > average]
    
    return average, maximum, minimum, highest_day, above_average

temperatures = np.array([32.5, 31.8, 30.2, 33.1, 34.5, 29.8, 31.0])

average, maximum, minimum, highest_day, above_average = temperatures_analysis(temperatures)

print("Average Temperature:", average)
print("Maximum Temperature:", maximum)
print("Minimum Temperature:", minimum)
print("Day with Highest Temperature:", highest_day)
print("Temperatures Above Average:", above_average)
```

### Problem 8. Scientific Data Transformation

Write your solution as a function that accepts an array as input.

```python
def scientific_data_transformation(array):
    squares = np.square(array)
    square_roots = np.sqrt(array)
    exponentials = np.exp(array)

    square_mean = np.mean(squares)
    square_root_mean = np.mean(square_roots)
    exponential_mean = np.mean(exponentials)

    return squares, square_roots, exponentials, square_mean, square_root_mean, exponential_mean


measurements = np.array([2, 4, 6, 8, 10, 12])

squares, square_roots, exponentials, square_mean, square_root_mean, exponential_mean = scientific_data_transformation(measurements)

print("Original Measurements:")
print(measurements)

print("\nSquare of Every Measurement:")
print(squares)

print("\nSquare Root of Every Measurement:")
print(square_roots)

print("\nExponential of Every Measurement:")
print(exponentials)

print("\nMean of Squares:")
print(square_mean)

print("\nMean of Square Roots:")
print(square_root_mean)

print("\nMean of Exponentials:")
print(exponential_mean)
```

### Problem 10. Data Combination

```python
import numpy as np

experiment_1 = np.array([10, 20, 30, 40, 50])
experiment_2 = np.array([15, 25, 35, 45, 55])

all_measurements = np.concatenate((experiment_1, experiment_2))

row_matrix = np.vstack((experiment_1, experiment_2))

column_matrix = np.column_stack((experiment_1, experiment_2))

print("Experiment 1:")
print(experiment_1)

print("\nExperiment 2:")
print(experiment_2)

print("\nAll Measurements:")
print(all_measurements)

print("Shape:", all_measurements.shape)

print("\nExperiments as Separate Rows:")
print(row_matrix)

print("Shape:", row_matrix.shape)

print("\nExperiments as Separate Columns:")
print(column_matrix)

print("Shape:", column_matrix.shape)
```

### Problem 15. Final Challenge: Student Performance Analysis
 Create a complete student performance analysis program. Consider: • 10 students • 4 subjects • Marks between 0 and 100 Generate the marks using NumPy. Your program should: (a) Store the marks in a two-dimensional NumPy array. (b) Calculate the average mark of every student. (c) Calculate the average mark of every subject. (d) Identify the student with the highest average. (e) Identify the student with the lowest average. (f) Identify the highest mark obtained in the entire dataset. (g) Identify the lowest mark obtained in the entire dataset. (h) Count how many students have an average above 75. (i) Extract all marks greater than 90. (j) Assign a grade to each student’s average using conditional statements. (k) Store the final student information in an appropriate Python data structure. (l) Write functions wherever appropriate. The final output should present the analysis in a clear and readable format.

```python
import numpy as np


def calculate_grade(average):
    if average >= 90:
        return "A"
    elif average >= 80:
        return "B"
    elif average >= 70:
        return "C"
    elif average >= 60:
        return "D"
    else:
        return "F"


def analyze_performance(marks):
    student_averages = np.mean(marks, axis=1)
    subject_averages = np.mean(marks, axis=0)

    highest_student = np.argmax(student_averages)
    lowest_student = np.argmin(student_averages)

    highest_mark = np.max(marks)
    lowest_mark = np.min(marks)

    students_above_75 = np.sum(student_averages > 75)

    marks_above_90 = marks[marks > 90]

    student_information = []

    for i in range(len(student_averages)):
        grade = calculate_grade(student_averages[i])

        student = {
            "student": i + 1,
            "average": student_averages[i],
            "grade": grade
        }

        student_information.append(student)

    return (
        student_averages,
        subject_averages,
        highest_student,
        lowest_student,
        highest_mark,
        lowest_mark,
        students_above_75,
        marks_above_90,
        student_information
    )


marks = np.random.randint(0, 101, (10, 4))

(
    student_averages,
    subject_averages,
    highest_student,
    lowest_student,
    highest_mark,
    lowest_mark,
    students_above_75,
    marks_above_90,
    student_information
) = analyze_performance(marks)


print("=" * 50)
print("       STUDENT PERFORMANCE ANALYSIS")
print("=" * 50)

print("\nMarks of Students:")
print(marks)

print("\nStudent Averages:")
for i, average in enumerate(student_averages):
    print(f"Student {i + 1}: {average:.2f}")

print("\nSubject Averages:")
for i, average in enumerate(subject_averages):
    print(f"Subject {i + 1}: {average:.2f}")

print("\nStudent with Highest Average:")
print(f"Student {highest_student + 1} - {student_averages[highest_student]:.2f}")

print("\nStudent with Lowest Average:")
print(f"Student {lowest_student + 1} - {student_averages[lowest_student]:.2f}")

print("\nHighest Mark in Entire Dataset:")
print(highest_mark)

print("\nLowest Mark in Entire Dataset:")
print(lowest_mark)

print("\nNumber of Students with Average Above 75:")
print(students_above_75)

print("\nAll Marks Greater Than 90:")
print(marks_above_90)

print("\nFinal Student Information:")
print("-" * 50)
print(f"{'Student':<10}{'Average':<15}{'Grade':<10}")
print("-" * 50)

for student in student_information:
    print(
        f"{student['student']:<10}"
        f"{student['average']:<15.2f}"
        f"{student['grade']:<10}"
    )

print("-" * 50)
```
