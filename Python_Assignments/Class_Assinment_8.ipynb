```python
class BankAccount:
    def __init__(self, account_num, owner, balance):
        self.account_num = account_num
        self.owner = owner
        self.balance = balance
    
    def deposit(self, amount):
        self.balance += amount
        print("Deposited:", amount)

    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance -= amount
        #     print("Withdrawn:", amount)
        # else:
        #     print("Insufficient balance")
            return True
        else:
            return False
            
    def display(self):
        print("Account Number:", self.account_num)
        print("Owner:", self.owner)
        print("Balance:", self.balance)
        
    def __str__(self):
        return (
            f"Account Number: {self.account_num}\n"
            f"Owner: {self.owner}\n"
            f"Balance: {self.balance}"
        )

account1 = BankAccount("101", "Pankaj", 1000)
account2 = BankAccount("102", "John", 2000)

print(account1.account_num, account1.owner, account1.balance)
print(account2.account_num, account2.owner, account2.balance)
```

```python
account1.deposit(500)
account1.display()
account1.withdraw(200)
account1.display()
account1.withdraw(3000)
```

```python
result = account1.withdraw(150)

print(f"Withdrawal Status: {result}")
print(f"Account Balance: {account1.balance}")
```

```python
result = account1.withdraw(5000)

print(f"Withdrawal Status: {result}")
print(f"Account Balance: {account1.balance}")
```

```python
class Student:
    def __init__(self, name, roll_no, marks):
        self.name = name
        self.roll_no = roll_no
        self.marks = marks
        
        # average_marks
    def average_marks(self):
        return sum(self.marks) / len(self.marks)
```

```python
Student1 = Student("Alice", 101, [85, 90, 78])
Student2 = Student("Bob", 102, [75, 80, 88])
Student3 = Student("Rahul", 103, [92, 88, 95])
```

```python
print(f"{Student1.name} (Roll No: {Student1.roll_no}) - Average Marks: {Student1.average_marks()}")
print(f"{Student2.name} (Roll No: {Student2.roll_no}) - Average Marks: {Student2.average_marks()}")
print(f"{Student3.name} (Roll No: {Student3.roll_no}) - Average Marks: {Student3.average_marks()}")
```

```python
class Vehicle:
    def __init__(self, brand):
        self.brand = brand
        
    def move(self):
        print("Vehicle is moving")


class Car(Vehicle):
    def __init__(self, brand, doors):
        super().__init__(brand)
        self.doors = doors
    
    def move(self):
        print("Car is driving on the road")


class Boat(Vehicle):
    def move(self):
        print("Boat is sailing on water")
```

```python
vehicle = Vehicle()
car = Car()
boat = Boat()
```

```python

vehicle.move()
car.move()
boat.move()
```

```python
car1 = Car("Toyota", 4)
car2 = Car("BMW", 2)
```

```python
print("Car 1:")
print("Brand:", car1.brand)
print("Doors:", car1.doors)
```

```python
print("Car 2:")
print("Brand:", car2.brand)
print("Doors:", car2.doors)
```

```python
class Car:
    def move(self):
        print("Car is driving on the road")


class Boat:
    def move(self):
        print("Boat is sailing on water")


class Airplane:
    def move(self):
        print("Airplane is flying in the sky")
```

```python

car = Car()
boat = Boat()
airplane = Airplane()

vehicles = [car, boat, airplane]
```

```python
for vehicle in vehicles:
    vehicle.move()
```
