```python
#Indexing and Slicing Lists
fruits = ["apple", "banana", "cherry", "date", "elderberry"]
print(fruits[0])  # Accessing the first element
print(fruits[-1])  # Accessing the last element
print(fruits[1:4])  # Slicing from index 1 to 3
```
