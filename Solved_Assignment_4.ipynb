### Membership operators(in/not in)

```python
fruits = ["apple", "banana", "cherry"]
print("apple" in fruits)
print("mango" not in fruits)
print("a" in "banana")
```

### Identity Operators(is/is not)

```python
a = [1, 2, 3]
b = [1, 2, 3]

print(a == b)
print(a is b)
print(a is not b)
```

### Enumerate

```python
items = ["pen", "Book", "Bag"]
for index, items in enumerate(items):
    print(index, items)
    
for index, items in enumerate(items, start=1):
    print(index, items)
    
```

### *args(uncertain number of parameters)

```python
def total(*numbers):
    return sum(numbers)

print(total(1, 2, 3))
print(total(5, 10))
```

### Positional, keyword and default arguments

```python
def intro(name, age, city="unknown"):
    print(f"{name} is {age}")
    
    intro("Alice")
    intro("Bob", 25)
    intro("charlie", 30, "New York")
```

### Q. write a function for its x is an number of parameters.

```python
def add_numbers(a, b, c):
    return a + b + c

result = add_numbers(10, 20, 30)
print("sum = ",result)
```

```python
def add_numbers(*numbers):
    total = 0
    
    for num in numbers:
        total += num
        
    return total

print(add_numbers(10,20,30))
print(add_numbers(1,2,3))
print(add_numbers(10,20,30,40,50))
```
