### isinstance()


```python
print(isinstance("alice", str))
print(isinstance("Alice", int))
print(isinstance(3.14, (int, float)))
```

### Membership operators(in/not in)

```python
fruits = ["apple", "banana", "cherry"]
print("apple" in fruits)
print("mango" not in fruits)
print("a" in "banana")
```

### Identity Operators(is/is not)

```python
a = [1, 2, 3]
b = [1, 2, 3]

print(a == b)
print(a is b)
print(a is not b)
```

### Enumerate

```python
items = ["pen", "Book", "Bag"]
for index, items in enumerate(items):
    print(index, items)
    
for index, items in enumerate(items, start=1):
    print(index, items)
    
```
